```python
import pandas as pd
import urllib
from sqlalchemy import create_engine

# 1. إعدادات خادم Docker
SERVER = '127.0.0.1,1433'  # أو 'localhost,1433'
DATABASE = 'AdventureWorks2022'
USERNAME = 'sa'
PASSWORD = 'Simo.1234'  # كلمة المرور التي حددتها عند إنشاء الحاوية

# 2. إعداد سلسلة الاتصال باستخدام المحرك المتاح 'SQL Server'
params = urllib.parse.quote_plus(
    f"DRIVER={{SQL Server}};"
    f"SERVER={SERVER};"
    f"DATABASE={DATABASE};"
    f"UID={USERNAME};"
    f"PWD={PASSWORD};"
)

# 3. إنشاء محرك SQLAlchemy
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

# 4. اختبر سحب البيانات
query = "SELECT TOP 5 * FROM Sales.SalesOrderHeader;"
df = pd.read_sql(query, engine)

print("تم الاتصال وسحب البيانات بنجاح:")
print(df.head())
```
