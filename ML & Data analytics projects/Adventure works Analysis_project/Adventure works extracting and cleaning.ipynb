```python
query6 = """SELECT 
                ISNULL(st.Name, 'Online Customer') AS Store_or_channel,
                -- p.Name AS Product,
                ISNULL(p2.Name, 'Uncategorized') AS Category,
                ISNULL(p1.Name, 'Uncategorized') AS SubCategory,
                ISNULL(p.Color, 'N/A') AS Color,
                p.StandardCost AS Cost,
                d.OrderQty AS Quantity,
                d.UnitPrice,
                d.UnitPriceDiscount,
                d.LineTotal AS Revenue,
                (d.LineTotal - (d.OrderQty * p.StandardCost)) AS Profit,
                YEAR(h.OrderDate) AS Order_year,
                MONTH(h.OrderDate) AS Order_month,
                DATENAME(MONTH, h.OrderDate) AS Order_month_name,
                DATENAME(WEEKDAY, h.OrderDate) AS Order_day_name,
                DATEDIFF(DAY, h.OrderDate, h.ShipDate) AS Process_days_to_ship,
                DATEDIFF(DAY, h.DueDate, h.ShipDate) AS Delivery_delay_days, 
                p.ReorderPoint AS Reorder_point

            FROM Sales.SalesOrderDetail d
            INNER JOIN Sales.SalesOrderHeader h 
                ON d.SalesOrderID = h.SalesOrderID
            INNER JOIN Production.Product p 
                ON d.ProductID = p.ProductID
            LEFT JOIN Production.ProductSubcategory p1 
                ON p.ProductSubcategoryID = p1.ProductSubcategoryID
            LEFT JOIN Production.ProductCategory p2 
                ON p1.ProductCategoryID = p2.ProductCategoryID
            LEFT JOIN Sales.Customer c 
                ON h.CustomerID = c.CustomerID
            LEFT JOIN Sales.Store st 
                ON c.StoreID = st.BusinessEntityID;
    """

df6 = pd.read_sql(query6, engine)
```
